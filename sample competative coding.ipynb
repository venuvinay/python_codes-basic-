```python
flip=int(input())
dicflip=int(input())
shipflip=int(input())
am=int(input())
diam=int(input())
sham=int(input())
snap=int(input())
snapdic=int(input())
shipsha=int(input())
dic1=flip*dicflip/100
flipkart=(flip-dic1)+shipflip
dic2=am*diam/100
amzon=(am-dic2)+sham
dic3=snap*snapdic/100
snapdeal=(snap-dic3)+shipsha
print("flipkart:",flipkart)
print("amzon:",amzon)
print("snapdeal:",snapdeal)
if flipkart<amzon and flipkart<snapdeal:
    print("He will prefer flipkart")
elif amzon<snapdeal:
    print("He will prefer amzon")
else:
    print("He will prefer snapdeal")
```

```python
height=float(input())
weight=float(input())
h=height/100
v=h*h
bmi=weight/v
if bmi<18.5:
    print("You are underweight. Have an apple daily")
else:
    if bmi>=18.5 and bmi<=24.9:
        print("You are normal. Go walking daily and maintain it")
    else:
        if bmi>=25.0 and bmi<=29.9:
            print("You are overweight. Go to Gym daily")
        else:
            print("You are obese. Go to doctor")
```
